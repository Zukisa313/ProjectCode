```python
import os
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis


# This script computes basic statistics for log returns of specified stocks

data_folder = "data"
stocks = ['AAPL', 'ADBE', 'AMAT', 'AMZN', 'CSCO', 'GOOGL', 'INTC', 
          'INTU', 'KLAC', 'LRCX', 'META', 'MSFT', 'MU', 'NFLX', 
          'NVDA', 'QCOM', 'TSLA', 'TXN']


results = []
for stock in stocks:
    file_path = os.path.join(data_folder, f"{stock}.csv")
    
    try:

        df = pd.read_csv(file_path)

        if 'Log_Return' not in df.columns:
            print(f" Skipping {stock}: 'Log_Return' column not found.")
            continue
        
        # Drop NaN values (just in case)
        returns = df['Log_Return'].dropna()
        
        # Compute descriptive statistics
        mean = returns.mean()
        std_dev = returns.std()
        skewness = skew(returns)
        kurt = kurtosis(returns, fisher=True)  
        min_val = returns.min()
        max_val = returns.max()
        data_range = max_val - min_val
        
        # Append to results list
        results.append({
            "Stock": stock,
            "Mean": mean,
            "Std_Dev": std_dev,
            "Skewness": skewness,
            "Kurtosis": kurt,
            "Min": min_val,
            "Max": max_val,
            "Range": data_range
        })
    
    except Exception as e:
        print(f"Error processing {stock}: {e}")


stats_df = pd.DataFrame(results)
stats_df = stats_df.round(6)

# Save to CSV
output_file = os.path.join(data_folder, "basic_statistics.csv")
stats_df.to_csv(output_file, index=False)

print(f"\n Basic statistics saved to: {output_file}")
print(stats_df)
```
