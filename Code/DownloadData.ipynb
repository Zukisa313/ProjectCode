```python
import yfinance as yf
import pandas as pd
import numpy as np
import os

def download_and_process_stocks(tickers, start_date, end_date):
    """
    Downloads historical stock data, computes log returns, and saves them to CSV files.

    Args:
        tickers (list): A list of stock ticker symbols.
        start_date (str): The start date for the data in 'YYYY-MM-DD' format.
        end_date (str): The end date for the data in 'YYYY-MM-DD' format.
    """
    # --- Create a subfolder for the data if it doesn't exist ---
    output_dir = 'data'
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        print(f"Created directory: '{output_dir}/'")

    for ticker in tickers:
        print(f"\n--- Processing {ticker} from {start_date} to {end_date} ---")
        try:
            # Download the historical data with adjusted prices
            stock_data = yf.download(ticker, start=start_date, end=end_date, progress=False, auto_adjust=True)

            if stock_data.empty:
                print(f"Could not download data for {ticker}. Please check the ticker and date range.")
                continue

            print(f"Data for {ticker} downloaded successfully. Calculating log returns...")

            # --- Ensure Close prices is a Series ---
            close_prices = stock_data['Close']
            if isinstance(close_prices, pd.DataFrame):
                close_prices = close_prices.iloc[:, 0]

            # --- Calculate Logarithmic Returns ---
            log_returns = np.log(close_prices / close_prices.shift(1)).dropna()

            # --- Create DataFrame for log returns ---
            returns_df = pd.DataFrame({
                'Date': log_returns.index,
                'Log_Return': log_returns.values
            })

            # --- Save the results to a CSV file ---
            output_path = os.path.join(output_dir, f"{ticker}.csv")
            returns_df.to_csv(output_path, index=False)

            print(f"Successfully saved log returns for {ticker} to '{output_path}'")
            print("\nFirst 5 rows of the saved data:")
            print(returns_df.head())

        except Exception as e:
            print(f"An error occurred while processing {ticker}: {e}")


def main():
    """
    Main function to define parameters and run the script.
    """
    # List of 20 NASDAQ-100 companies that have been listed for at least 10 years
    tickers_to_process = [
        'AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA', 'META', 'NVDA', 'PYPL', 'CSCO', 'INTC',
        'AMD', 'NFLX', 'ADBE', 'INTU', 'QCOM', 'TXN', 'AMAT', 'MU', 'KLAC', 'LRCX'
    ]
    
    # Define the 10-year period
    end_date = '2024-12-31'
    start_date = '2014-12-31'
    
    download_and_process_stocks(tickers_to_process, start_date, end_date)


if __name__ == "__main__":
    main()

```
